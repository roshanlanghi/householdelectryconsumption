### 📌 Objective
Load the raw UCI Individual Household Electric Power Consumption dataset and perform initial structural inspection.



```python
import os
import pandas as pd
import numpy as np

RAW_DATA_PATH = os.path.join("..", "data", "raw", "household_power_consumption.txt")
print(f"Path to raw data: {RAW_DATA_PATH}")
print(f"File exists: {os.path.exists(RAW_DATA_PATH)}")

```

### 📥 Load Dataset
The dataset is semi-colon separated (`;`) and missing values are denoted by `'?'`.



```python
# Load the dataset
# Dataset size is ~2M records; low_memory=False is recommended
df = pd.read_csv(
    RAW_DATA_PATH,
    sep=';',
    na_values='?',
    low_memory=False
)

print(f"Dataset Shape: {df.shape[0]:,} rows and {df.shape[1]} columns")

```

### 📊 Missing Value Overview


```python
missing_counts = df.isnull().sum()
missing_pct = (missing_counts / len(df)) * 100
missing_df = pd.DataFrame({'Missing Values': missing_counts, 'Percentage (%)': missing_pct})
missing_df
```

### 📝 Summary of Initial Observations
1. The dataset contains 2,075,259 records collected at 1-minute intervals between Dec 2006 and Nov 2010.
2. Approximately 1.25% of values are missing (`?` parsed as `NaN`).
3. Date and Time are separated strings requiring conversion to a consolidated `Datetime` index.
4. Numerical features will need type conversion and preprocessing.
