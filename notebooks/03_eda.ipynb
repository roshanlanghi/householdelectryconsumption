### 📌 Objective
Analyze data distributions, daily and monthly consumption patterns, sub-metering breakdown, and correlation relationships.



```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

PROCESSED_DATA_PATH = os.path.join("..", "data", "processed", "cleaned_data.csv")
df = pd.read_csv(PROCESSED_DATA_PATH)
df['Datetime'] = pd.to_datetime(df['Datetime'])
df.info()

```

### 1. Statistical Summary


```python
df.describe().T
```

### 2. Target Variable Distribution: Global Active Power


```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Global_active_power'], kde=True, color='#2563EB', bins=60)
plt.title("Distribution of Global Active Power (kW)", fontsize=14, fontweight='bold')
plt.xlabel("Global Active Power (kW)")
plt.ylabel("Frequency")
plt.show()

```

### 3. Consumption Trend Across Hours of the Day


```python
hourly_avg = df.groupby('Hour')['Global_active_power'].mean().reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(data=hourly_avg, x='Hour', y='Global_active_power', marker='o', color='#7C3AED', linewidth=2.5)
plt.title("Average Electricity Consumption by Hour of Day", fontsize=14, fontweight='bold')
plt.xlabel("Hour (0 - 23)")
plt.ylabel("Average Active Power (kW)")
plt.xticks(range(0, 24))
plt.show()

```

### 4. Sub-Metering Breakdown
- **Sub_metering_1:** Kitchen (dishwasher, microwave, oven)
- **Sub_metering_2:** Laundry room (washing machine, tumble-dryer, refrigerator, light)
- **Sub_metering_3:** Climate control (electric water heater and air conditioner)



```python
sub_metering_avg = df[['Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']].mean()

plt.figure(figsize=(8, 5))
sub_metering_avg.plot(kind='bar', color=['#06B6D4', '#F59E0B', '#10B981'])
plt.title("Average Sub-Metering Consumption (Watt-hours)", fontsize=14, fontweight='bold')
plt.ylabel("Watt-hours")
plt.xticks(rotation=0)
plt.show()

```

### 5. Feature Correlation Heatmap


```python
features = [
    'Global_active_power', 'Global_reactive_power', 'Voltage',
    'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3',
    'Hour', 'Day', 'Month', 'DayOfWeek'
]

plt.figure(figsize=(10, 8))
corr = df[features].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', cbar=True, square=True)
plt.title("Correlation Heatmap", fontsize=14, fontweight='bold')
plt.show()

```

### 💡 Key Findings
1. **Strong Collinearity:** `Global_active_power` and `Global_intensity` show near-perfect linear correlation (~0.99), as current intensity is directly tied to power consumption.
2. **Peak Consumption:** Highest power usage occurs between 19:00 and 22:00 in the evening.
3. **Sub-metering 3** (Water heating & AC) dominates household energy consumption.
